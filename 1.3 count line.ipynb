```python
import os

def count_total_csv_rows(folder_path):
    global_total_rows = 0

    # 遍历目标文件夹及其所有子文件夹
    # root 代表当前正在遍历的目录地址
    # dirs 代表该目录下的子目录列表
    # files 代表该目录下的文件列表
    for root, dirs, files in os.walk(folder_path):
        subfolder_total_rows = 0
        csv_count = 0  # 记录当前子文件夹中的csv数量
        
        for file in files:
            # 筛选出所有的 csv 文件
            if file.endswith('.csv'):
                csv_count += 1
                file_path = os.path.join(root, file)
                
                # 实时打印正在处理的文件
                print(f"正在处理: {file_path}")
                
                try:
                    # 读取文件（使用 ignore 忽略编码乱码问题）
                    # 如果数据是gbk编码，可以将 utf-8 改为 gbk
                    with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
                        line_count = sum(1 for _ in f)
                        
                        # 如果文件有内容，减去1行表头；如果是空文件，则记为0
                        data_rows = line_count - 1 if line_count > 0 else 0
                        
                    subfolder_total_rows += data_rows
                        
                except Exception as e:
                    print(f"  [错误] 读取文件时发生异常 {file_path}: {e}")

        # 如果当前子文件夹下有csv文件，打印该子文件夹的汇总信息
        if csv_count > 0:
            print(f">>> 【子文件夹汇总】: 文件夹 [{root}] 处理完毕")
            print(f">>> 包含CSV数量: {csv_count} 个")
            print(f">>> 该子文件夹总行数(排除表头): {subfolder_total_rows} 行")
            print("=" * 60)
            
        # 累加到全局总行数
        global_total_rows += subfolder_total_rows

    # 最终打印所有文件的总计
    print(f"\n【全部统计完成】")
    print(f"目标文件夹: {folder_path}")
    print(f"所有CSV总行数(排除表头)合计为: {global_total_rows} 行")

if __name__ == "__main__":
    # 指定你的文件夹路径
    target_folder = r"E:\202512LLM交通\data\1.裁判文书网数据"
    
    # 检查路径是否存在
    if os.path.exists(target_folder):
        print(f"开始统计文件夹：{target_folder}\n" + "="*60)
        count_total_csv_rows(target_folder)
    else:
        print(f"找不到文件夹路径：{target_folder}，请检查路径是否正确。")
```
