# 合并所有的csv

```python
import pandas as pd

# ================= 自定义文件路径 =================
# 在这里填写你的 4 个源 CSV 文件的具体路径
file1 = '../data/7.匹配时间/time_transport_result_1_4.csv'
file2 = '../data/7.匹配时间/time_transport_result_2_4.csv'
file3 = '../data/7.匹配时间/time_transport_result_3_4.csv'
file4 = '../data/7.匹配时间/time_transport_result_4_4.csv'

# 在这里填写合并后新 CSV 文件的保存路径
output_file = '../data/7.匹配时间/merged_output.csv'
# =================================================

def merge_csv_files():
    # 将文件路径放入一个列表中
    input_files = [file1, file2, file3, file4]
    
    # 创建一个空列表，用于存放读取的数据
    dataframes = []
    
    print("开始读取文件...")
    for file in input_files:
        try:
            # 读取单个 CSV 文件
            df = pd.read_csv(file)
            dataframes.append(df)
            print(f"成功读取: {file}")
        except Exception as e:
            print(f"读取文件 {file} 时出错: {e}")
            return # 如果报错就停止运行

    print("正在拼接数据...")
    # 使用 concat 将所有表拼接在一起
    # ignore_index=True 会重新生成连续的行索引，防止原来的索引冲突
    merged_df = pd.concat(dataframes, ignore_index=True)

    print("正在保存文件...")
    # 保存为新的 CSV 文件
    # index=False 表示保存时不包含最前面的数字索引列
    # encoding='utf-8-sig' 能够防止中文字符在 Excel 中打开时乱码
    merged_df.to_csv(output_file, index=False, encoding='utf-8-sig')
    
    print(f"合并完成！总共合并了 {len(merged_df)} 行数据。")
    print(f"文件已成功保存至: {output_file}")

if __name__ == "__main__":
    merge_csv_files()
```
